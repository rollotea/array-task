```python
from pathlib import Path
import sys

sys.path.append(str(Path.cwd().parent))

from src.gen_trace import gen_trace
from src.utils import ArrayTaskLoader


json_files = list(Path("../datasets/training/raw/v1").glob("*.json"))

loader = ArrayTaskLoader()

tasks = []
for json_file in json_files:
    task = loader.load(json_file)
    tasks.append(task)


traces = gen_trace(tasks)

results = []
for trace in traces:
    accepted_rule = [i for i in trace if i["accepted"] == True]
    if len(accepted_rule) == 1:
        results.append({"task_id": accepted_rule[0]["task_id"], "trace": accepted_rule[0]["trace"]})
    else:
        results.append({"task_id": min(accepted_rule, key=len)["task_id"], "trace": min(accepted_rule, key=len)["trace"]})

import json
import os

def output_trace(
    version_name: str,
    results,
):
    os.makedirs(f"../traces/{version_name}", exist_ok=True)
    for result in results:
        with open(
            # f"test/{dir}/{version_name}/{rule['id']}_{i}.json",
            f"../traces/{version_name}/{result['task_id']}.json",
            "w",
            encoding="utf-8",
        ) as f:
            json.dump(
                result,
                f,
                # indent=2,
                ensure_ascii=False,
            )

output_trace("single_candidate_v2", results)
```

# 正解と不正解のルール両方

```python
from pathlib import Path
import sys
import random

sys.path.append(str(Path.cwd().parent))

from src.gen_trace import gen_trace
from src.utils import ArrayTaskLoader


json_files = list(Path("../datasets/training/raw/v2").glob("*.json"))

loader = ArrayTaskLoader()

tasks = []
for json_file in json_files:
    task = loader.load(json_file)
    tasks.append(task)


traces = gen_trace(tasks)

results = []
for trace in traces:
    accepted_rule = [i for i in trace if i["accepted"] == True]
    lines = random.choices([varify["trace"] for varify in trace if varify["accepted"] == False], k=random.randint(0, 4))
    if len(accepted_rule) == 1:
        lines.append(accepted_rule[0]["trace"])
        results.append({"task_id": accepted_rule[0]["task_id"], "trace": "\n".join(lines)})
    else:
        lines.append(min(accepted_rule, key=len)["trace"])
        results.append({"task_id": min(accepted_rule, key=len)["task_id"], "trace": "\n".join(lines)})

import json
import os

def output_trace(
    version_name: str,
    results,
):
    os.makedirs(f"../traces/{version_name}", exist_ok=True)
    for result in results:
        with open(
            # f"test/{dir}/{version_name}/{rule['id']}_{i}.json",
            f"../traces/{version_name}/{result['task_id']}.json",
            "w",
            encoding="utf-8",
        ) as f:
            json.dump(
                result,
                f,
                # indent=2,
                ensure_ascii=False,
            )

output_trace("multi_candidate_v2", results)
```

## dataset確認

```python
from src.utils import ArrayTaskLoader, PromptBuilder


json_files = list(Path("../datasets/training/raw/v1").glob("*.json"))

loader = ArrayTaskLoader()

tasks = []
for json_file in json_files:
    task = loader.load(json_file)
    tasks.append(task)

builder = PromptBuilder()

for task in tasks:
    prompt = builder.build(task)
    print("===" * 50)
    print(f"task.task_id: {task.task_id}")
    print("===" * 50)
    print(prompt.encode().decode("unicode_escape"))
    print(task.test_case.output_array)
```

```python
import json

from src.utils import Trace

class TraceLoader:
    def load(self, path: Path) -> Trace:
        with open(path, "r") as f:
            data = json.load(f)

        return Trace(task_id=data["task_id"], trace=data["trace"])

trace_json_files = list(Path("../traces/v1").glob("*.json"))
trace_loader = TraceLoader()
traces = []
for json_file in trace_json_files:
    trace = trace_loader.load(json_file)
    traces.append(trace)

for task in tasks:
    task.trace = next(trace.trace for trace in traces if trace.task_id == task.task_id)
```

```python
for task in tasks:
    prompt = builder.build(task)
    print("###" * 50)
    print(f"task.task_id: {task.task_id}")
    print("===" * 50)
    print(prompt.encode().decode("unicode_escape"))
    print(task.test_case.output_array)
    print("===" * 50)
    print(task.trace)
```

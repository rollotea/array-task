```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent.parent))
from src.gen_trace import PRIMITIVES, apply_rule


from itertools import permutations

from src.utils import ArrayTaskLoader, PromptBuilder


json_files = list(Path("../datasets/training/raw/v3").glob("*.json"))

loader = ArrayTaskLoader()

tasks = []
for json_file in json_files:
    task = loader.load(json_file)
    tasks.append(task)
primitive_names = list(PRIMITIVES.keys())


rules = []

for name in primitive_names:
    rules.append([name])

for rule in permutations(primitive_names, 2):
    rules.append(list(rule))

def solve(task):
    solutions = []

    for rule in rules:
        if all(
            apply_rule(example.input_array, rule) == example.output_array
            for example in task.examples
        ):
            solutions.append(rule)

    return solutions

for task_id, task in enumerate(tasks):
    if (task_id + 1) % 50 == 0:
        solutions = solve(task)

        if len(solutions) == 0:
            print(f"[UNSOLVED] task_{task.task_id}")
        elif len(solutions) == 1:
            pass
            print(f"[OK] task_{task.task_id}: {solutions[0]}")
        else:
            print(f"[AMBIGUOUS] task_{task.task_id}: {solutions}")
```

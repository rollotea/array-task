- https://aclanthology.org/2024.naacl-srw.3.pdf

### 論文その9
- Rethinking Generalization in Reasoning SFT: A Conditional Analysis on Optimization, Data, and Model Capability
- https://arxiv.org/abs/2604.06628
- SFTは記憶で、強化学習は汎化を促す見方が一般的、だけど場合によってはsftでも汎化が起きる
- 学習に使用するデータやベースのモデルなどの要因を変えながら、検証する
- 学習したものと似た問題だけでなく、学習の範囲外の問題も含めてテスト。数学の推論の問題を使う
- わかったこと
    - 汎化の性能は一時的に低下してから向上する。学習不足
    - データの質も重要。推論の後戻りや検証の手順も学習させるのが大事
    - 元のモデルの能力も大事
- データセットの作り方
    - OpenR1-Math-220kっていう問題集からサンプリング
    - その問題をqwen3 32bに解かせて正解したものの推論過程を学習データとして使う

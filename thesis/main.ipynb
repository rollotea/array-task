# 先行研究

### 論文その1
- Exploring Compositional Generalization of Large Language Models
- https://aclanthology.org/2024.naacl-srw.3.pdf
- 自然言語でルールを指示する
- 単純な命令を組み合わせる能力がLLMの分布外汎化である
- 多い命令の数で学習させて、少ない命令の数でテストしたら精度が上がったけど、逆は起こらなかった
- 例えば、「ジャスティンビーバーが生まれた年に大統領だったのはだれか」という質問は「ジャスティンビーバーが生まれた年」と「そのときの大統領」の二つを処理しないといけない
- 「翻訳してから要約して」というのもおなじ
- データセットの作り方
    - belleコーパスという英語の対話のデータセットから、chatGPTにお願いして問題をいくつか抽出してもらう
    - そのあとchatGPTに命令を付け足してもらう
- 学習
    - full fine tuningとLoRA
- 評価指標
    - BLEU
        - 単語の適合率
    - ROGUE
        - 文章が順序を保ったままどれだけ一致してるか
- 限界
    - 本研究では、構成的指示（compositional instructions）に対するLLMの汎化能力を調査します。しかし、言語の複雑さや解釈の多様性ゆえに、「構成的指示」という概念を厳密に定義することは困難です。例えば、「夏に関する記事を書いてください」という指示は、「記事を書く」と「その記事は夏に関するものであるべきだ」という要素に分解可能です。
    - 言語の複雑性のため、compositional instructionを正確に定義するのは難しい

### 論文その２
- Chain-of-Instructions: Compositional Instruction Tuning on Large Language Models
- https://ojs.aaai.org/index.php/AAAI/article/view/34574
- こちらも自然言語で表現された複数の処理を必要とされるタスクを作って実験した
- 「フランス語でブログ風のタイトルを生成せよ」
    - 与えられたテキストのタイトルを生成する
    - タイトルのスタイルをブログ記事のタイトル風に変更する
    - そのブログ記事のタイトルをフランス語に翻訳する
- 処理が１つだけ必要な問題から処理が３つ必要な問題までを作成し、それを生成AIに学習をさせる
- 処理が１つだけの問題をCoI1で2つならCoI2といった感じ
- 学習に使うのはCoI1, CoI2, CoI3, CoI12, CoI123で評価するのはCoI4, CoI5の問題
- CoI12とCoI123はCoI4とCoI5の問題を解くときに精度が上昇した。
- 多くの処理が必要な問題を学習させると、処理の数がそれ以上に多い問題を解くのが得意になる。

### 論文その３
- Rethinking Generalization in Reasoning SFT: A Conditional Analysis on Optimization, Data, and Model Capability
- https://arxiv.org/pdf/2604.06628
- 推論の過程を利用したSFT(自分の研究で扱う学習の方法)は一般的に問題のパターンを記憶するだけで、違う領域の問題を解決する能力にはつながらないという見方が一般的
- 算数の問題を使った学習をしても、コードを書く能力や科学に関する問題を解くような能力は育たないと考えられている
- しかし、特定の条件下ではSFTでも領域を超えて推論能力を向上させることができた。
    - 十分な学習を行う
        - 学習の初期では性能が下がるが、学習を継続すると性能は回復・向上する
    - 質のいいデータを使うこと
        - 質の低いデータでは、汎化能力を失う。そのため、検証済みの長い推論過程を使うことが大切
    - 能力の高いモデルを使用すること
        - 性能の低いモデルは推論過程を表面的に真似をするだけになってしまう
- これからの研究はSFTによって生成AIの推論能力が汎化するかどうかではなく、どんな条件で汎化するのかを調べることになる
- 小規模のモデルに対して、推論過程を表面的に真似をするだけになってしまい、汎化できていないということは述べている

# 中間発表で右側に書くもの

### 【研究目的/問題設定】

近年、大規模言語モデル（LLM）の性能向上において、教師ありファインチューニング（SFT）が広く用いられている。SFTによってモデルが学習データに含まれる問題を解けるようになるだけでなく、未知の問題に対する推論能力が汎化する可能性も指摘されている。一方で、SFTによってモデルが具体的にどのような処理能力を獲得し、それを未知の問題にどのように適用するようになるのかについては、十分に明らかになっていない。そこで本研究では、単純な処理を組み合わせた問題を用いて、SFT前後のLLMの問題解決能力を比較し、SFTによるモデルの変化を分析する。

### 【研究方法/研究経過】

ARC-AGIのような、入力と出力の例から変換規則を推測する問題に着想を得て、複数の単純な処理を組み合わせた問題を自作した。個々の処理を明示的に定義することで、問題を構成する処理や処理の組み合わせを研究者側で制御できるようにした。作成した問題を用いてQwen3に対してSFTを実施し、SFT前後の正解率を比較するとともに、生成された回答や処理過程を分析することで、SFTによるモデルの変化を調査する。
現在、答えのみを使用したSFTおよび推論過程を含めたSFTを実施し、テスト用の問題に対する評価を行っている。答えのみを使用したSFTでは正解率の向上が確認された一方、推論過程を含めたSFTでは正解率の向上が限定的であった。そのため現在は、正解率だけでは捉えられないSFTによる問題解決過程の変化について分析を進めている。
